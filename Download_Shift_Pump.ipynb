### Read and Cast data

```python
# Find all params.json files under pump_metadata
json_paths = sorted(glob.glob(os.path.join(METADATA_DIR, "sample_*", "params.json")))

# Load all json data into a list of dicts
records = []
for path in json_paths:
    with open(path, "r") as f:
        data = json.load(f)
        # Add the folder name as sample_id
        data["sample_id"] = os.path.basename(os.path.dirname(path))
        records.append(data)

# Create polars DataFrame
df = pl.DataFrame(records)

# Define column groupings for typing
float_cols = [
    "flow_rate",
    "head",
    "outer_diameter_factor",
    "outlet_width_factor",
    "shroud_diameter_factor",
    "volute_q_factor",
    "volute_r_factor",
    "teAngleDelta",
    "leAngleDelta",
    "compactness",
    "leWidthVar",
    "leRake",
    "teRake",
    "flow_rate_op_condition",
    "leShapevar",
    "shrAngleVar",
    "stream_velocity",
    "area_ref",
    "length_ref",
    "length_ref_pitch",
    "length_ref_roll",
    "length_ref_yaw",
    "moment_center_x",
    "moment_center_y",
    "moment_center_z",
    "R",
    "gamma",
    "cp",
    "air_density",
]

int_cols = [
    "rotational_speed",
    "fix_blade_number",
    "leAngleVar",
    "leShapeVar",
    "tipGap",
    "leEllipseFactor",
    "mach",
    "pressure",
    "temperature",
]

bool_cols = ["teTangential"]

# Apply casting expressions
cast_exprs = []

for col in float_cols:
    if col in df.columns:
        cast_exprs.append(pl.col(col).cast(pl.Float64))

for col in int_cols:
    if col in df.columns:
        cast_exprs.append(pl.col(col).cast(pl.Int64))

for col in bool_cols:
    if col in df.columns:
        # Handle string "true"/"false" safely
        if df[col].dtype == pl.String:
            cast_exprs.append(
                (pl.col(col).str.to_lowercase() == "true").cast(pl.Boolean)
            )
        else:
            cast_exprs.append(pl.col(col).cast(pl.Boolean))

df = df.with_columns(cast_exprs)
```

### Applying Kennard-Stone to select most distinct geometries

```python
# 1. Define the geometric columns to include in the shape analysis
active_geometric_cols = [
    "outer_diameter_factor",
    "outlet_width_factor",
    "shroud_diameter_factor",
    "teAngleDelta",
    "leAngleDelta",
    "compactness",
    "leWidthVar",
    "leRake",
    "teRake",
    "leShapevar",
]

# Ensure we only use columns that actually exist in the DataFrame
existing_geo_cols = [col for col in active_geometric_cols if col in df.columns]

# 2. Extract features, fill missing/null values, and cast booleans to floats
df_features = (
    df.select(existing_geo_cols)
    .with_columns(
        [
            # Cast boolean columns (like teTangential) to floats so they can be scaled
            pl.col(col).cast(pl.Float64)
            for col in existing_geo_cols
        ]
    )
    .fill_null(0.0)
)

X = df_features.to_numpy()

# 3. Scale features (essential so that features with larger ranges don't dominate the distance)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 4. Use the Kennard-Stone algorithm to select a diverse subset
# Set train_size to the number of shapes you want to download/encoder-fine-tune
num_to_select = 200

X_train, X_test, df_selected, df_remaining = ks.train_test_split(
    X_scaled, df, train_size=num_to_select
)

# 5. Output the selected sample folder names
selected_folders = df_selected["sample_id"].to_list()

print(f"Selected {len(selected_folders)} diverse shapes:")
print(selected_folders)
```

```python
df_filtered = df_selected.filter(pl.col("sample_id").is_in(selected_folders))
df_filtered.select(active_geometric_cols).describe()
```
